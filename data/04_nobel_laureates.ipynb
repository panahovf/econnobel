# Step 4 · Past winners of the Nobel Prize in Economics

**What this notebook does:** downloads the list of everyone who has won the prize (since 1969) from the
Nobel Prize's official data service, [api.nobelprize.org](https://api.nobelprize.org). Saves `nobel_laureates.csv`.

**Why we need it:** past winners stay on the poll's list, so people can find them, but they can't be picked.
The website shows "Won in 2024 · not eligible" next to their name.

Run it again once a new prize is announced. It takes a second.

## Setup

```python
import warnings

# The Mac's built-in Python prints a harmless warning about its SSL library when `requests` loads.
# Hide it. (This line has to come before `import requests`.)
warnings.filterwarnings("ignore")

import pandas as pd
import requests
```

## 1. Download the list

`nobelPrizeCategory=eco` asks for Economic Sciences only. `limit=200` makes sure all of them come in one answer
(there are about 100).

```python
URL = "https://api.nobelprize.org/2.1/laureates"
params = {"nobelPrizeCategory": "eco", "limit": 200}

# The Nobel Prize's server refuses requests that don't say who is asking, so we say who we are.
HEADERS = {"User-Agent": "Mozilla/5.0 (compatible; EconNobel poll; https://panahovf.github.io/econnobel/)"}

response = requests.get(URL, params=params, headers=HEADERS, timeout=60)
response.raise_for_status()   # stop here if the download failed
data = response.json()

print(len(data["laureates"]), "laureates")
```

## 2. One row per winner

Each laureate has a `knownName` (the name they are usually known by, e.g. "Daron Acemoglu") and a list of
their prizes with the year.

```python
rows = []

for person in data["laureates"]:
    name = person["knownName"]["en"]                 # the English version of their name
    for prize in person["nobelPrizes"]:
        rows.append({"name": name, "year": int(prize["awardYear"])})

laureates = pd.DataFrame(rows).sort_values(["year", "name"], ascending=[False, True])
laureates.head(10)
```

## 3. Save

```python
laureates.to_csv("nobel_laureates.csv", index=False)
print("Saved", len(laureates), "laureates to nobel_laureates.csv")
```
